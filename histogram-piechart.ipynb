```python
# histogram 

#Histogram एक ऐसा graph है जो बताता है कि किसी dataset में values कितनी बार (frequency) आती हैं।


import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_excel(r'R:\Visual Studio\GenAI Course\Matplotlib\students.xlsx',
)

import matplotlib.pyplot as plt

marks = [12, 18, 22, 25, 28, 35, 38, 42, 45, 48]

plt.hist(marks, bins=4, edgecolor='black')

plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Students Marks Histogram')

plt.show()
```

```python
# Pie Chart एक circular graph होता है, जो data को अलग-अलग हिस्सों (slices) में दिखाता है।
# इसका इस्तेमाल percentage या proportion दिखाने के लिए किया जाता है।

import matplotlib.pyplot as plt

subjects = ['Hindi', 'English', 'Math', 'Science']
marks = [25, 20, 35, 20]

plt.pie(marks, labels=subjects, autopct='%1.1f%%')

plt.title('Students Marks Pie Chart')

plt.show()
```

```python
#dynamic pie chart
import matplotlib.pyplot as plt

subjects = ['Math', 'History', 'Physics', 'Chemistry', 'Biology', 'English', 'Geography']
marks = [
    data['math_score'].sum(),
    data['history_score'].sum(),
    data['physics_score'].sum(),
    data['chemistry_score'].sum(),
    data['biology_score'].sum(),
    data['english_score'].sum(),
    data['geography_score'].sum()
]

plt.pie(marks, labels=subjects, autopct='%1.1f%%', startangle=90, explode=[0, 0, 0.5, 0, 0, 0, 0],shadow=True)

plt.title('Students Marks Pie Chart')
plt.legend(subjects, loc='upper left',
           bbox_to_anchor=(-0.6, 1))
plt.show()
```
