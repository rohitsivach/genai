```python
# pandas DataFrame

# DataFrame Pandas का सबसे महत्वपूर्ण data structure है।

# यह 2-Dimensional (Rows × Columns) data structure होता है, बिल्कुल Excel Sheet या SQL Table की तरह।

# Rows → Records
# Columns → Fields


#List dataframe

import pandas as pd

data = ([1,2,3],[4,5,6],[7,8,9])
dataframe = pd.DataFrame(data, columns=['col1', 'col2', 'col3'], index=['row1', 'row2', 'row3'])
print(dataframe)
```
