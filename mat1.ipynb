```python
#scatterplot
import matplotlib.pyplot as plt

age = [18, 20, 22, 25, 27]
salary = [20000, 25000, 30000, 40000, 45000]

age2 = [19, 21, 24, 26, 28]
salary2 = [22000, 28000, 35000, 42000, 50000]

plt.scatter(age, salary, color="blue", label="Group A")
plt.scatter(age2, salary2, color="red", label="Group B")

plt.xlabel("Age")
plt.ylabel("Salary")
plt.title("Age vs Salary")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

marks = data['math_score'].head(5)
marks2 = data['history_score'].head(5)

x = marks.index
y = marks.values
x2 = marks2.index
y2 = marks2.values

plt.scatter(
    x,
    y,
    color="blue",
    s=100,          # point size
    alpha=0.7       # transparency
)
plt.scatter(
    x2,
    y2,
    color="red",
    s=100,          # point size
    alpha=0.7       # transparency
)

plt.xlabel("Index")
plt.ylabel("Math/history Score")
plt.title("Top 5 Math Scores")
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt

subjects = ['Math', 'History', 'Physics', 'Chemistry', 'Biology', 'English', 'Geography']

marks = [
    data['math_score'].mean(),
    data['history_score'].mean(),
    data['physics_score'].mean(),
    data['chemistry_score'].mean(),
    data['biology_score'].mean(),
    data['english_score'].mean(),
    data['geography_score'].mean()
]

plt.bar(subjects, marks)

plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks by Subject")

plt.xticks(rotation=45)
plt.show()
```
