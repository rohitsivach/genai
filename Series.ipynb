```python
# Loops in Python

# Loop का उपयोग किसी block of code को बार-बार (repeatedly) चलाने के लिए किया जाता है।

# 1. for Loop

# जब आपको पहले से पता हो कि loop कितनी बार चलेगा, तब for loop का उपयोग करते हैं।

fruits = avg.unit_price

for fruit in avg.product_category:
    print(fruit)

fruits[fruits>200]
```

```python
num = 5

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
#Loop को तुरंत रोक देता है।
for i in range(1, 11):
    if i == 6:
        break
    print(i)

    
```

```python
#Current iteration को छोड़कर अगली iteration पर चला जाता है।
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
# 2. while Loop

# जब आपको नहीं पता कि loop कितनी बार चलेगा, और किसी condition के True रहने तक code चलाना हो, तब while loop का उपयोग करते हैं।

i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
count = 5

while count > 0:
    print(count)
    count -= 1

print("Done")
```
