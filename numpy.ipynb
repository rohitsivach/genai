```python
#Ye functions Data Wrangling, Data Cleaning, Pandas, aur Machine Learning preprocessing me bahut frequently use hote hain.

import numpy as np

arr = np.array([10,20,20,30,40])

print("Unique:", np.unique(arr))

print("Append:", np.append(arr,50))

print("Insert:", np.insert(arr,2,25))

print("Delete:", np.delete(arr,1))

print("Where:", np.where(arr>20))

print("Argmax:", np.argmax(arr))

print("Argmin:", np.argmin(arr))

a = np.array([1,2,3])
b = np.array([4,5,6])

print("Concatenate:", np.concatenate((a,b)))

print("Split:", np.split(np.array([10,20,30,40,50,60]),3))
```
